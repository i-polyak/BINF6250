1. “Is log monotonic? What should happen to the max in log space?” 
Yes, log is monotonic: if a > b, then log a > log b. So in log space, max stays max. Only the multiplications turn into additions. Durbin’s log-space formula confirms this (p. 77). 
The project text’s sentence about converting the max “into a sum of their log probabilities” contradicts this. 

2. “What is log 0?” 
Python’s math.log(0) crashes. NumPy’s np.log(0) returns -inf with a warning. 
We need to pick one way of handling zero probabilities. 
The instructions suggest pseudocounts as one option.

```python
def viterbi(self, observations):
        """Find the single most probable sequence of hidden states.

        Args:
            observations (str): observed symbols, e.g. "GGCACTGAA"
        Returns:
            (path, log_probability)
                path: list of states, one per observation
                log_probability: log of that path's probability
        """

#Viterbi algorithm:
#1. Start (first column). For each state, the score is that state’s initial probability times the chance that state emits the first letter. There’s nothing earlier to point back to yet.
#2. Fill the remaining columns, left to right. For each state in the new column:
    #Look at every box in the previous column and compute its score × the chance of moving from that state to this one.
    #Keep the biggest result, and write down which state it came from. That note is the pointer.
    #Multiply the biggest result by the chance this state emits this column’s letter.
#3. Finish. In the last column, the box with the biggest score holds the best path’s probability, and its row is the last state.
#4. Trace back. Follow the pointers backward to the first column, then reverse the list. That’s the answer.

#Log space: take the log of every probability once at the start, and then every multiplication becomes addition. 
    #Durbin says Viterbi “should always be done in log space” (p. 56), and recommends taking the logs of all the model parameters before running, rather than inside the loop (p. 78). 
    #This prevents long sequences from shrinking to 0.0.
```
