```python
class HMM:
    """Hidden Markov model built from three probability dictionaries."""

    def __init__(self, init_probs, trans_probs, emit_probs):
        """
        Args:
            init_probs  (dict): {state: probability of starting in that state}
            trans_probs (dict): {from_state: {to_state: probability}}
            emit_probs  (dict): {state: {symbol: probability}}
        """
```

```python
def viterbi(self, observations):
        """Find the single most probable sequence of hidden states.

        Args:
            observations (str): observed symbols, e.g. "GGCACTGAA"
        Returns:
            (path, log_probability)
                path: list of states, one per observation
                log_probability: log of that path's probability
        """
```
